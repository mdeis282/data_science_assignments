```python
#1
def quicksort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quicksort(left) + middle + quicksort(right)
```

```python
#2
def nested_loop_example(matrix):
    rows, cols = len(matrix), len(matrix[0])
    total = 0
    for i in range(rows):
        for j in range(cols):
            total += matrix[i][j]
    return total
```

```python
#4
def longest_increasing_subsequence(nums):
    n = len(nums)
    lis = [1] * n
    for i in range(1, n):
        for j in range(0, i):
            if nums[i] > nums[j] and lis[i] < lis[j] + 1:
                lis[i] = lis[j] + 1
    return max(lis)
```

```python
#5
def mysterious_function(arr):
    n = len(arr)
    result = 0
    for i in range(n):
        for j in range(i, n):
            result += arr[i] * arr[j]
    return result
```

```python
#6
def sum_of_digits(n):
    if n == 0:
        return 0
    return n % 10 + sum_of_digits(n // 10)

# Example usage
n = 123
print(f"The sum of digits of {n} is: {sum_of_digits(n)}")
```

```python
#7
def fibonacci_series(n):
    if n == 0:
        return [0]
    elif n == 1:
        return [0, 1]
    else:
        series = fibonacci_series(n - 1)
        series.append(series[-1] + series[-2])
        return series

# Example usage
n = 6
print(f"The first {n} numbers of the Fibonacci series are: {fibonacci_series(n)}")
```

```python
#8
def subset_sum(nums, target, index=0):
    if target == 0:
        return True
    if index >= len(nums) or target < 0:
        return False
    # Include the number at the current index or exclude it
    return subset_sum(nums, target - nums[index], index + 1) or subset_sum(nums, target, index + 1)

# Example usage
nums = [3, 34, 4, 12, 5, 2]
target = 9
print(f"Is there a subset that adds up to {target}? {subset_sum(nums, target)}")
```

```python
#9
def word_break(s, word_dict):
    if not s:
        return True
    for word in word_dict:
        if s.startswith(word):
            if word_break(s[len(word):], word_dict):
                return True
    return False

# Example usage
s = "leetcode"
word_dict = ["leet", "code"]
print(f"Can the string '{s}' be segmented into a sequence of dictionary words? {word_break(s, word_dict)}")
```

```python
#10
def solve_n_queens(n):
    def is_safe(board, row, col):
        # Check this column on the left side
        for i in range(col):
            if board[row][i] == 'Q':
                return False
        # Check upper diagonal on the left side
        for i, j in zip(range(row, -1, -1), range(col, -1, -1)):
            if board[i][j] == 'Q':
                return False
        # Check lower diagonal on the left side
        for i, j in zip(range(row, n), range(col, -1, -1)):
            if board[i][j] == 'Q':
                return False
        return True

    def solve(board, col):
        if col >= n:
            result.append([''.join(row) for row in board])
            return
        for i in range(n):
            if is_safe(board, i, col):
                board[i][col] = 'Q'
                solve(board, col + 1)
                board[i][col] = '.'

    result = []
    board = [['.' for _ in range(n)] for _ in range(n)]
    solve(board, 0)
    return result

# Example usage
n = 4
solutions = solve_n_queens(n)
print(f"Solutions for {n}-Queens problem:")
for solution in solutions:
    for row in solution:
        print(row)
    print()
```
