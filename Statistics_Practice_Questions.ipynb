```python
#(i)
import random

int_list = [random.randint(90, 130) for _ in range(100)]
```

```python
#(ii)
def calculate_mean(numbers):
    return sum(numbers) / len(numbers)

def calculate_median(numbers):
    sorted_numbers = sorted(numbers)
    n = len(sorted_numbers)
    if n % 2 == 0:
        return (sorted_numbers[n//2 - 1] + sorted_numbers[n//2]) / 2
    else:
        return sorted_numbers[n//2]

mean = calculate_mean(int_list)
median = calculate_median(int_list)
```

```python
#(iii)
from collections import Counter

def calculate_mode(numbers):
    count = Counter(numbers)
    max_freq = max(count.values())
    mode = [k for k, v in count.items() if v == max_freq]
    return mode

mode = calculate_mode(int_list)
```

```python
#(iv)
def calculate_weighted_mean(values, weights):
    weighted_sum = sum(v * w for v, w in zip(values, weights))
    total_weight = sum(weights)
    return weighted_sum / total_weight

# Example: Let's assume some weights
weights = [random.random() for _ in range(100)]
weighted_mean = calculate_weighted_mean(int_list, weights)
```

```python
#(v)
import math

def calculate_geometric_mean(numbers):
    product = 1
    for number in numbers:
        product *= number
    return product ** (1/len(numbers))

geometric_mean = calculate_geometric_mean(int_list)
```

```python
#(vi)
def calculate_harmonic_mean(numbers):
    return len(numbers) / sum(1/x for x in numbers)

harmonic_mean = calculate_harmonic_mean(int_list)
```

```python
#(vii)
def calculate_midrange(numbers):
    return (min(numbers) + max(numbers)) / 2

midrange = calculate_midrange(int_list)
```

```python
#(viii)
def calculate_trimmed_mean(numbers, percentage):
    trimmed_amount = int(len(numbers) * percentage)
    trimmed_list = sorted(numbers)[trimmed_amount:-trimmed_amount]
    return sum(trimmed_list) / len(trimmed_list)

trimmed_mean = calculate_trimmed_mean(int_list, 0.1)  # Example: 10% trimmed mean
```

```python
class DiscreteRandomVariable:
    def __init__(self, values, probabilities):
        """
        Initialize the discrete random variable.
        :param values: A list of possible values.
        :param probabilities: A list of corresponding probabilities for the values.
        """
        if len(values) != len(probabilities):
            raise ValueError("Values and probabilities lists must be of the same length.")

        if not all(0 <= p <= 1 for p in probabilities):
            raise ValueError("All probabilities must be between 0 and 1.")

        if sum(probabilities) != 1:
            raise ValueError("The sum of the probabilities must be 1.")

        self.values = values
        self.probabilities = probabilities

    def expected_value(self):
        """
        Calculate the expected value (mean) of the random variable.
        :return: The expected value.
        """
        return sum(v * p for v, p in zip(self.values, self.probabilities))

    def variance(self):
        """
        Calculate the variance of the random variable.
        :return: The variance.
        """
        mean = self.expected_value()
        return sum(p * ((v - mean) ** 2) for v, p in zip(self.values, self.probabilities))

# Example usage
values = [1, 2, 3, 4, 5]
probabilities = [0.1, 0.2, 0.3, 0.2, 0.2]

rv = DiscreteRandomVariable(values, probabilities)
expected_value = rv.expected_value()
variance = rv.variance()

print(f"Expected Value: {expected_value}")
print(f"Variance: {variance}")
```

```python
import random

class FairDieSimulation:
    def __init__(self):
        self.outcomes = [1, 2, 3, 4, 5, 6]
        self.probabilities = [1/6] * 6

    def expected_value(self):
        """
        Calculate the expected value of rolling the die.
        :return: The expected value.
        """
        return sum(v * p for v, p in zip(self.outcomes, self.probabilities))

    def variance(self):
        """
        Calculate the variance of rolling the die.
        :return: The variance.
        """
        mean = self.expected_value()
        return sum(p * ((v - mean) ** 2) for v, p in zip(self.outcomes, self.probabilities))

# Simulate the rolling of the die
simulation = FairDieSimulation()
expected_value = simulation.expected_value()
variance = simulation.variance()

print(f"Expected Value of the Die Roll: {expected_value}")
print(f"Variance of the Die Roll: {variance}")
```

```python
import numpy as np

def generate_samples(distribution_type, **kwargs):
    """
    Generate random samples from a specified distribution and calculate their mean and variance.

    :param distribution_type: Type of distribution ('binomial', 'poisson').
    :param kwargs: Parameters specific to the distribution type.
    :return: Mean and Variance of the generated samples.
    """
    if distribution_type == 'binomial':
        # Example: n=10, p=0.5, size=1000
        samples = np.random.binomial(n=kwargs['n'], p=kwargs['p'], size=kwargs['size'])
    elif distribution_type == 'poisson':
        # Example: lam=5, size=1000
        samples = np.random.poisson(lam=kwargs['lam'], size=kwargs['size'])
    else:
        raise ValueError("Unsupported distribution type. Please use 'binomial' or 'poisson'.")

    mean = np.mean(samples)
    variance = np.var(samples)

    return mean, variance

# Example usage for Binomial distribution
binomial_mean, binomial_variance = generate_samples(
    distribution_type='binomial', n=10, p=0.5, size=1000)

# Example usage for Poisson distribution
poisson_mean, poisson_variance = generate_samples(
    distribution_type='poisson', lam=5, size=1000)

print(f"Binomial Distribution -> Mean: {binomial_mean}, Variance: {binomial_variance}")
print(f"Poisson Distribution -> Mean: {poisson_mean}, Variance: {poisson_variance}")
```

```python
import numpy as np

def generate_gaussian_samples(mean, std_dev, size):
    """
    Generate random numbers from a Gaussian (normal) distribution.

    :param mean: Mean of the distribution.
    :param std_dev: Standard deviation of the distribution.
    :param size: Number of samples to generate.
    :return: Generated samples, mean, variance, and standard deviation of the samples.
    """
    samples = np.random.normal(loc=mean, scale=std_dev, size=size)
    calculated_mean = np.mean(samples)
    variance = np.var(samples)
    std_deviation = np.std(samples)

    return samples, calculated_mean, variance, std_deviation

# Example usage
mean = 0   # Mean of the Gaussian distribution
std_dev = 1  # Standard deviation of the Gaussian distribution
size = 1000  # Number of samples

samples, calculated_mean, variance, std_deviation = generate_gaussian_samples(mean, std_dev, size)

print(f"Generated Mean: {calculated_mean}")
print(f"Generated Variance: {variance}")
print(f"Generated Standard Deviation: {std_deviation}")
```

```python
#(ii)
def calculate_skewness(column):
    """
    Calculate the skewness of a given column.

    :param column: Pandas Series representing the column.
    :return: Skewness of the column.
    """
    return column.skew()

# Calculate skewness for 'total_bill' and 'tip' columns
total_bill_skewness = calculate_skewness(tips['total_bill'])
tip_skewness = calculate_skewness(tips['tip'])

print(f"Skewness of 'total_bill': {total_bill_skewness}")
print(f"Skewness of 'tip': {tip_skewness}")
```

```python
#(iii)
def skewness_type(skewness_value):
    """
    Determine the type of skewness.

    :param skewness_value: The skewness value of the column.
    :return: Type of skewness.
    """
    if skewness_value > 0:
        return "Positive Skewness"
    elif skewness_value < 0:
        return "Negative Skewness"
    else:
        return "Approximately Symmetric"

total_bill_skewness_type = skewness_type(total_bill_skewness)
tip_skewness_type = skewness_type(tip_skewness)

print(f"Total Bill Column: {total_bill_skewness_type}")
print(f"Tip Column: {tip_skewness_type}")
```

```python
#(iv)
def calculate_covariance(column1, column2):
    """
    Calculate the covariance between two columns.

    :param column1: Pandas Series representing the first column.
    :param column2: Pandas Series representing the second column.
    :return: Covariance between the two columns.
    """
    return np.cov(column1, column2)[0, 1]

# Calculate covariance between 'total_bill' and 'tip'
covariance_value = calculate_covariance(tips['total_bill'], tips['tip'])
print(f"Covariance between 'total_bill' and 'tip': {covariance_value}")
```

```python
#(v)
def calculate_pearson_correlation(column1, column2):
    """
    Calculate the Pearson correlation coefficient between two columns.

    :param column1: Pandas Series representing the first column.
    :param column2: Pandas Series representing the second column.
    :return: Pearson correlation coefficient.
    """
    return column1.corr(column2)

# Calculate Pearson correlation coefficient between 'total_bill' and 'tip'
pearson_correlation = calculate_pearson_correlation(tips['total_bill'], tips['tip'])
print(f"Pearson Correlation between 'total_bill' and 'tip': {pearson_correlation}")
```

```python
import math

def normal_pdf(x, mean, std_dev):
    """
    Calculate the probability density function (PDF) for a given normal distribution.

    :param x: The value at which to evaluate the PDF.
    :param mean: The mean (mu) of the normal distribution.
    :param std_dev: The standard deviation (sigma) of the normal distribution.
    :return: The value of the PDF at x.
    """
    exponent = math.exp(-((x - mean) ** 2) / (2 * std_dev ** 2))
    coefficient = 1 / (std_dev * math.sqrt(2 * math.pi))
    return coefficient * exponent

# Example usage
mean = 0     # Mean of the normal distribution
std_dev = 1  # Standard deviation of the normal distribution
x_value = 1  # Value at which to evaluate the PDF

pdf_value = normal_pdf(x_value, mean, std_dev)
print(f"PDF value at x={x_value} for N({mean}, {std_dev}^2): {pdf_value}")
```

```python
import math

def exponential_cdf(x, rate):
    """
    Calculate the cumulative distribution function (CDF) for an exponential distribution.

    :param x: The value at which to evaluate the CDF.
    :param rate: The rate parameter (lambda) of the exponential distribution.
    :return: The value of the CDF at x.
    """
    if x < 0:
        return 0.0
    else:
        return 1 - math.exp(-rate * x)

# Example usage
rate = 0.5  # Rate parameter (lambda) of the exponential distribution
x_value = 2  # Value at which to evaluate the CDF

cdf_value = exponential_cdf(x_value, rate)
print(f"CDF value at x={x_value} for exponential distribution with rate={rate}: {cdf_value}")
```

```python
import math

def poisson_pmf(k, lam):
    """
    Calculate the probability mass function (PMF) for a Poisson distribution.

    :param k: The number of occurrences (an integer).
    :param lam: The rate parameter (lambda) of the Poisson distribution.
    :return: The value of the PMF at k.
    """
    return (math.exp(-lam) * lam**k) / math.factorial(k)

# Example usage
lam = 3  # Rate parameter (lambda) of the Poisson distribution
k_value = 2  # Number of occurrences

pmf_value = poisson_pmf(k_value, lam)
print(f"PMF value at k={k_value} for Poisson distribution with lambda={lam}: {pmf_value}")
```

```python
import numpy as np

# 50 purchases out of 1000 visitors for the old layout
old_layout = np.array([1] * 50 + [0] * 950)

# 70 purchases out of 1000 visitors for the new layout
new_layout = np.array([1] * 70 + [0] * 930)
```

```python
from statsmodels.stats.proportion import proportions_ztest

# Calculate the number of successes and the number of trials
successes = np.array([old_layout.sum(), new_layout.sum()])
nobs = np.array([len(old_layout), len(new_layout)])

# Perform the z-test
z_stat, p_value = proportions_ztest(successes, nobs)

# Output the results
print(f"Z-statistic: {z_stat}")
print(f"P-value: {p_value}")

# Interpret the results
if p_value < 0.05:
    print("Reject the null hypothesis: There is a significant difference in conversion rates.")
    if z_stat > 0:
        print("The new layout has a significantly higher conversion rate.")
    else:
        print("The old layout has a significantly higher conversion rate.")
else:
    print("Fail to reject the null hypothesis: There is no significant difference in conversion rates.")
```

```python
import numpy as np

layout_a_clicks = np.array([28, 32, 33, 29, 31, 34, 30, 35, 36, 37])
layout_b_clicks = np.array([40, 41, 38, 42, 39, 44, 43, 41, 45, 47])
```

```python
from scipy import stats

def ab_test_analysis(clicks_a, clicks_b):
    # Calculate the t-statistic and p-value for a two-sample t-test
    t_stat, p_value = stats.ttest_ind(clicks_a, clicks_b)

    # Calculate the degrees of freedom
    n1 = len(clicks_a)
    n2 = len(clicks_b)
    dof = n1 + n2 - 2

    # Output the results
    print(f"T-statistic: {t_stat}")
    print(f"Degrees of Freedom: {dof}")
    print(f"P-value: {p_value}")

    # Interpret the results
    if p_value < 0.05:
        print("Reject the null hypothesis: There is a significant difference in click-through rates between the two layouts.")
        if t_stat > 0:
            print("Layout B has a significantly higher click-through rate.")
        else:
            print("Layout A has a significantly higher click-through rate.")
    else:
        print("Fail to reject the null hypothesis: There is no significant difference in click-through rates between the two layouts.")

# Run the A/B test analysis
ab_test_analysis(layout_a_clicks, layout_b_clicks)
```

```python
import numpy as np
import pandas as pd
from scipy import stats

# Set seed for reproducibility
np.random.seed(0)

# Generate age groups and voter preferences
age_groups = np.random.choice(['18-30', '31-50', '51+', '51+'], size=500)
voter_preferences = np.random.choice(['Candidate A', 'Candidate B'], size=500)
```

```python
# Create a DataFrame to organize the data
data = pd.DataFrame({'Age Group': age_groups, 'Voter Preference': voter_preferences})

# Create a contingency table
contingency_table = pd.crosstab(data['Age Group'], data['Voter Preference'])

print("Contingency Table:")
print(contingency_table)
```

```python
# Perform the Chi-Square test of independence
chi2_stat, p_value, dof, expected = stats.chi2_contingency(contingency_table)

# Output the results
print(f"Chi-Square Statistic: {chi2_stat}")
print(f"P-value: {p_value}")
print(f"Degrees of Freedom: {dof}")
print("Expected Frequencies:")
print(expected)

# Interpret the results
if p_value < 0.05:
    print("Reject the null hypothesis: There is a significant association between age groups and voter preferences.")
else:
    print("Fail to reject the null hypothesis: There is no significant association between age groups and voter preferences.")
```

```python
import numpy as np
from scipy import stats

# Sample data: Product satisfaction levels (rows) vs. Customer regions (columns)
data = np.array([[50, 30, 40, 20],
                 [30, 40, 30, 50],
                 [20, 30, 40, 30]])
```

```python
def chi_square_test(contingency_table):
    # Perform the Chi-Square test of independence
    chi2_stat, p_value, dof, expected = stats.chi2_contingency(contingency_table)

    # Output the results
    print(f"Chi-Square Statistic: {chi2_stat}")
    print(f"P-value: {p_value}")
    print(f"Degrees of Freedom: {dof}")
    print("Expected Frequencies:")
    print(expected)

    # Interpret the results
    if p_value < 0.05:
        print("Reject the null hypothesis: There is a significant relationship between product satisfaction levels and customer regions.")
    else:
        print("Fail to reject the null hypothesis: There is no significant relationship between product satisfaction levels and customer regions.")

# Run the Chi-Square test
chi_square_test(data)
```

```python
import numpy as np
from scipy import stats

# Sample data: Customer satisfaction scores for each product version
standard_scores = np.array([80, 85, 90, 78, 88, 82, 92, 78, 85, 87])
premium_scores = np.array([90, 92, 88, 92, 95, 91, 96, 93, 89, 93])
deluxe_scores = np.array([95, 98, 92, 97, 96, 94, 98, 97, 92, 99])
```

```python
def perform_anova(*groups):
    # Perform the one-way ANOVA test
    f_stat, p_value = stats.f_oneway(*groups)

    # Output the results
    print(f"F-statistic: {f_stat}")
    print(f"P-value: {p_value}")

    # Interpret the results
    if p_value < 0.05:
        print("Reject the null hypothesis: There is a significant difference in customer satisfaction scores among the product versions.")
    else:
        print("Fail to reject the null hypothesis: There is no significant difference in customer satisfaction scores among the product versions.")

# Run the ANOVA test
perform_anova(standard_scores, premium_scores, deluxe_scores)
```
