```python
#1
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load your dataset (update the path if needed)
df = pd.read_csv('/content/heart_failure_clinical_records_dataset - heart_failure_clinical_records_dataset.csv')

# Set the style for seaborn
sns.set(style="whitegrid")

# Distribution of Age
plt.figure(figsize=(10, 6))
sns.histplot(df['age'], bins=20, kde=True, color='skyblue')
plt.title('Distribution of Age among Heart Failure Patients')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
#2
# Death Rate Variation with Age
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='age', hue='DEATH_EVENT', multiple='stack', bins=20, palette='viridis')
plt.title('Death Rate Variation with Age')
plt.xlabel('Age')
plt.ylabel('Number of Patients')
plt.show()
```

```python
#3
# Percentage of Male and Female Patients
gender_counts = df['sex'].value_counts(normalize=True) * 100
gender_counts.plot(kind='bar', color=['lightblue', 'lightgreen'], figsize=(8, 6))
plt.title('Percentage of Male and Female Patients')
plt.xlabel('Sex (0 = Female, 1 = Male)')
plt.ylabel('Percentage')
plt.show()
```

```python
#4
# Platelet Count Variation among Age Groups
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='age', y='platelets', hue='DEATH_EVENT', palette='viridis')
plt.title('Platelet Count Variation among Different Age Groups')
plt.xlabel('Age')
plt.ylabel('Platelet Count')
plt.show()
```

```python
#5
# Correlation between Creatinine and Sodium Levels
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='serum_creatinine', y='serum_sodium', hue='DEATH_EVENT', palette='viridis')
plt.title('Correlation between Serum Creatinine and Sodium Levels')
plt.xlabel('Serum Creatinine')
plt.ylabel('Serum Sodium')
plt.show()

# Calculate the correlation coefficient
correlation = df['serum_creatinine'].corr(df['serum_sodium'])
print(f'Correlation coefficient between serum creatinine and serum sodium: {correlation}')
```

```python
#6
# Prevalence of High Blood Pressure by Gender
plt.figure(figsize=(8, 6))
sns.countplot(data=df, x='sex', hue='high_blood_pressure', palette='viridis')
plt.title('Prevalence of High Blood Pressure by Gender')
plt.xlabel('Sex (0 = Female, 1 = Male)')
plt.ylabel('Count')
plt.show()
```

```python
#7
# Smoking Habits and Heart Failure
plt.figure(figsize=(8, 6))
sns.countplot(data=df, x='smoking', hue='DEATH_EVENT', palette='viridis')
plt.title('Relationship between Smoking Habits and Occurrence of Heart Failure')
plt.xlabel('Smoking (0 = Non-Smoker, 1 = Smoker)')
plt.ylabel('Count')
plt.show()
```

```python
#8
# Distribution of Death Events across Different Age Groups
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='age', hue='DEATH_EVENT', multiple='stack', bins=20, palette='viridis')
plt.title('Distribution of Death Events across Different Age Groups')
plt.xlabel('Age')
plt.ylabel('Number of Patients')
plt.show()
```

```python
#9
# Ejection Fraction and Diabetes
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='diabetes', y='ejection_fraction', palette='viridis')
plt.title('Ejection Fraction by Diabetes Status')
plt.xlabel('Diabetes (0 = No, 1 = Yes)')
plt.ylabel('Ejection Fraction')
plt.show()

# Calculate the statistical significance
from scipy.stats import ttest_ind

no_diabetes = df[df['diabetes'] == 0]['ejection_fraction']
with_diabetes = df[df['diabetes'] == 1]['ejection_fraction']

t_stat, p_value = ttest_ind(no_diabetes, with_diabetes)
print(f'T-test statistic: {t_stat}, p-value: {p_value}')
```

```python
#10
# Serum Creatinine and Survival
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='DEATH_EVENT', y='serum_creatinine', palette='viridis')
plt.title('Serum Creatinine Levels by Survival Status')
plt.xlabel('Survival (0 = Survived, 1 = Died)')
plt.ylabel('Serum Creatinine')
plt.show()

# Calculate the statistical significance
from scipy.stats import ttest_ind

survived = df[df['DEATH_EVENT'] == 0]['serum_creatinine']
died = df[df['DEATH_EVENT'] == 1]['serum_creatinine']

t_stat, p_value = ttest_ind(survived, died)
print(f'T-test statistic: {t_stat}, p-value: {p_value}')
```
