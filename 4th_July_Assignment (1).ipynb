````python
name = ["freedomFighter", "Bapuji", "Mohan", "dash", "karam", "chandra", "gandhi"]
length1 = len((name[-len(name)+1:-1:2]))
length2 = len((name[-len(name)+1:-1:1]))
print(length1 + length2)
````

name[-len(name)+1:-1:2] slices the list starting from index 1 to the second last element, with a step of 2.

name[-len(name)+1:-1:1] slices the list starting from index 1 to the second last element, with a step of 1.

Calculating these:

````python
name = ["freedomFighter", "Bapuji", "Mohan", "dash", "karam", "chandra", "gandhi"]

slice1 = name[-len(name)+1:-1:2]  # ['Bapuji', 'dash', 'chandra']
length1 = len(slice1)

slice2 = name[-len(name)+1:-1:1]  # ['Bapuji', 'Mohan', 'dash', 'karam', 'chandra']
length2 = len(slice2)

print(length1 + length2)  # 3 + 5 = 8
````

Explanation:

This expression returns x if the condition is True, otherwise it returns y.

Nested Ternary Operators: We can nest ternary operators to handle multiple conditions:

First, we compare a with both b and c. If a is larger than both, a is the largest.

If not, we then compare b and c to find the larger of the two.

````python
def calculate_grade(percentage):
    match percentage:
        case _ if percentage > 85:
            return 'A'
        case _ if 75 <= percentage <= 85:
            return 'B'
        case _ if 50 <= percentage < 75:
            return 'C'
        case _ if 30 < percentage < 50:
            return 'D'
        case _ if percentage <= 30:
            return 'Reappear'

def main():
    marks = []

    # Accepting marks of the student in five major subjects
    for i in range(1, 6):
        mark = float(input(f"Enter the marks for subject {i}: "))
        marks.append(mark)

    # Displaying the marks
    print("\nMarks entered:")
    for i, mark in enumerate(marks, 1):
        print(f"Subject {i}: {mark}")

    # Calculating the sum of the marks
    total_marks = sum(marks)

    # Calculating the percentage
    percentage = total_marks / 5

    # Displaying the total marks and percentage
    print(f"\nTotal Marks: {total_marks}")
    print(f"Percentage: {percentage}%")

    # Finding the grade
    grade = calculate_grade(percentage)
    print(f"Grade: {grade}")

if __name__ == "__main__":
    main()
````

````python
def determine_color(wavelength):
    if 400.0 <= wavelength <= 440.0:
        return "Violet"
    elif 440.0 < wavelength <= 460.0:
        return "Indigo"
    elif 460.0 < wavelength <= 500.0:
        return "Blue"
    elif 500.0 < wavelength <= 570.0:
        return "Green"
    elif 570.0 < wavelength <= 590.0:
        return "Yellow"
    elif 590.0 < wavelength <= 620.0:
        return "Orange"
    elif 620.0 < wavelength <= 720.0:
        return "Red"
    else:
        return "Wavelength out of range"

def main():
    try:
        wavelength = float(input("Enter the wavelength (in nm): "))
        color = determine_color(wavelength)
        print(f"The color corresponding to the wavelength {wavelength} nm is {color}.")
    except ValueError:
        print("Please enter a valid numerical value for the wavelength.")

if __name__ == "__main__":
    main()
````

````python
class LibraryBook:
    def __init__(self, book_name, author, available=True):
        self.__book_name = book_name
        self.__author = author
        self.__available = available

    # Getter methods
    def get_book_name(self):
        return self.__book_name

    def get_author(self):
        return self.__author

    def is_available(self):
        return self.__available

    # Method for borrowing a book
    def borrow_book(self):
        if self.__available:
            self.__available = False
            print(f"You have successfully borrowed '{self.__book_name}' by {self.__author}.")
        else:
            print(f"Sorry, '{self.__book_name}' by {self.__author} is currently not available.")

    # Method for returning a book
    def return_book(self):
        if not self.__available:
            self.__available = True
            print(f"Thank you for returning '{self.__book_name}' by {self.__author}.")
        else:
            print(f"'{self.__book_name}' by {self.__author} was not borrowed.")

    # Method to display book information
    def display_info(self):
        status = "Available" if self.__available else "Not Available"
        print(f"Book Name: {self.__book_name}")
        print(f"Author: {self.__author}")
        print(f"Status: {status}")

# Test the borrowing and returning functionality with sample data
def main():
    book1 = LibraryBook("1984", "George Orwell")
    book2 = LibraryBook("To Kill a Mockingbird", "Harper Lee")

    # Display initial information
    print("Initial information of book1:")
    book1.display_info()

    print("\nInitial information of book2:")
    book2.display_info()

    # Borrow book1
    print("\nBorrowing book1...")
    book1.borrow_book()

    # Try borrowing book1 again
    print("\nTrying to borrow book1 again...")
    book1.borrow_book()

    # Return book1
    print("\nReturning book1...")
    book1.return_book()

    # Try returning book1 again
    print("\nTrying to return book1 again...")
    book1.return_book()

    # Display updated information
    print("\nUpdated information of book1:")
    book1.display_info()

    print("\nUpdated information of book2:")
    book2.display_info()

if __name__ == "__main__":
    main()
````

````python
class BankAccount:
    def __init__(self, account_number, balance=0.0):
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited ${amount:.2f}. New balance: ${self.balance:.2f}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew ${amount:.2f}. New balance: ${self.balance:.2f}")
        else:
            print("Withdrawal amount must be positive and less than or equal to the balance.")

    def get_balance(self):
        return self.balance

    def display_info(self):
        print(f"Account Number: {self.account_number}")
        print(f"Balance: ${self.balance:.2f}")

class SavingsAccount(BankAccount):
    def __init__(self, account_number, balance=0.0, interest_rate=0.01):
        super().__init__(account_number, balance)
        self.interest_rate = interest_rate

    def apply_interest(self):
        interest = self.balance * self.interest_rate
        self.balance += interest
        print(f"Applied interest of ${interest:.2f}. New balance: ${self.balance:.2f}")

class CheckingAccount(BankAccount):
    def __init__(self, account_number, balance=0.0, overdraft_limit=100.0):
        super().__init__(account_number, balance)
        self.overdraft_limit = overdraft_limit

    def withdraw(self, amount):
        if 0 < amount <= (self.balance + self.overdraft_limit):
            self.balance -= amount
            print(f"Withdrew ${amount:.2f}. New balance: ${self.balance:.2f}")
        else:
            print("Withdrawal amount exceeds overdraft limit.")

def main():
    # Create instances of SavingsAccount and CheckingAccount
    savings = SavingsAccount("S123", 1000.0, 0.05)
    checking = CheckingAccount("C456", 500.0, 200.0)

    # Display initial information
    print("Savings Account Info:")
    savings.display_info()
    print("\nChecking Account Info:")
    checking.display_info()

    # Perform transactions
    print("\nDepositing $200 into savings account...")
    savings.deposit(200)

    print("\nWithdrawing $300 from checking account...")
    checking.withdraw(300)

    print("\nApplying interest to savings account...")
    savings.apply_interest()

    print("\nAttempting to withdraw $800 from checking account...")
    checking.withdraw(800)

    # Display updated information
    print("\nUpdated Savings Account Info:")
    savings.display_info()
    print("\nUpdated Checking Account Info:")
    checking.display_info()

if __name__ == "__main__":
    main()
````

````python
class MenuItem:
    _id_counter = 1  # Class-level attribute for unique ID generation

    def __init__(self, name, description, price, category):
        self.__id = MenuItem._id_counter
        MenuItem._id_counter += 1
        self.name = name
        self.description = description
        self.price = price
        self.category = category

    # Getter for ID
    def get_id(self):
        return self.__id

    # Method to update item information
    def update_item(self, name=None, description=None, price=None, category=None):
        if name is not None:
            self.name = name
        if description is not None:
            self.description = description
        if price is not None:
            self.price = price
        if category is not None:
            self.category = category

    # Method to display item information
    def display_info(self):
        print(f"ID: {self.get_id()}")
        print(f"Name: {self.name}")
        print(f"Description: {self.description}")
        print(f"Price: ${self.price:.2f}")
        print(f"Category: {self.category}")

class FoodItem(MenuItem):
    def __init__(self, name, description, price, category, cuisine_type):
        super().__init__(name, description, price, category)
        self.cuisine_type = cuisine_type

    # Method to display food item information
    def display_info(self):
        super().display_info()
        print(f"Cuisine Type: {self.cuisine_type}")

class BeverageItem(MenuItem):
    def __init__(self, name, description, price, category, is_alcoholic):
        super().__init__(name, description, price, category)
        self.is_alcoholic = is_alcoholic

    # Method to display beverage item information
    def display_info(self):
        super().display_info()
        print(f"Alcoholic: {'Yes' if self.is_alcoholic else 'No'}")

# Method to add a new item
def add_item(menu, item):
    menu.append(item)

# Method to remove an item by ID
def remove_item(menu, item_id):
    for item in menu:
        if item.get_id() == item_id:
            menu.remove(item)
            print(f"Item with ID {item_id} removed from the menu.")
            return
    print(f"Item with ID {item_id} not found.")

# Method to update an item by ID
def update_item(menu, item_id, **kwargs):
    for item in menu:
        if item.get_id() == item_id:
            item.update_item(**kwargs)
            print(f"Item with ID {item_id} updated.")
            return
    print(f"Item with ID {item_id} not found.")

def main():
    menu = []

    # Adding items to the menu
    food1 = FoodItem("Pizza", "Cheese Pizza", 10.99, "Food", "Italian")
    beverage1 = BeverageItem("Coke", "Coca Cola", 1.99, "Beverage", False)
    add_item(menu, food1)
    add_item(menu, beverage1)

    # Displaying menu items
    print("\nMenu Items:")
    for item in menu:
        item.display_info()
        print()

    # Updating an item
    print("Updating the price of Pizza...")
    update_item(menu, food1.get_id(), price=12.99)

    # Removing an item
    print("Removing Coke from the menu...")
    remove_item(menu, beverage1.get_id())

    # Displaying updated menu items
    print("\nUpdated Menu Items:")
    for item in menu:
        item.display_info()
        print()

if __name__ == "__main__":
    main()
````

````python
class Room:
    _id_counter = 1  # Class-level attribute for unique ID generation

    def __init__(self, room_number, room_type, rate):
        self.__id = Room._id_counter
        Room._id_counter += 1
        self.room_number = room_number
        self.room_type = room_type
        self.rate = rate
        self.__available = True

    # Getter for ID
    def get_id(self):
        return self.__id

    # Method to book a room
    def book_room(self):
        if self.__available:
            self.__available = False
            print(f"Room {self.room_number} has been booked.")
        else:
            print(f"Room {self.room_number} is already booked.")

    # Method to check in a guest
    def check_in(self):
        if not self.__available:
            print(f"Guest checked into room {self.room_number}.")
        else:
            print(f"Room {self.room_number} is not booked yet. Please book the room first.")

    # Method to check out a guest
    def check_out(self):
        if not self.__available:
            self.__available = True
            print(f"Guest checked out of room {self.room_number}.")
        else:
            print(f"Room {self.room_number} is already available.")

    # Method to display room information
    def display_info(self):
        status = "Available" if self.__available else "Not Available"
        print(f"Room ID: {self.get_id()}")
        print(f"Room Number: {self.room_number}")
        print(f"Room Type: {self.room_type}")
        print(f"Rate: ${self.rate:.2f}")
        print(f"Status: {status}")

class SuiteRoom(Room):
    def __init__(self, room_number, rate, additional_services):
        super().__init__(room_number, "Suite", rate)
        self.additional_services = additional_services

    # Method to display suite room information
    def display_info(self):
        super().display_info()
        print(f"Additional Services: {self.additional_services}")

class StandardRoom(Room):
    def __init__(self, room_number, rate, bed_type):
        super().__init__(room_number, "Standard", rate)
        self.bed_type = bed_type

    # Method to display standard room information
    def display_info(self):
        super().display_info()
        print(f"Bed Type: {self.bed_type}")

def main():
    # Create instances of SuiteRoom and StandardRoom
    suite = SuiteRoom(101, 200.0, "Breakfast, WiFi")
    standard = StandardRoom(102, 100.0, "Queen Size Bed")

    # Display initial information
    print("Suite Room Info:")
    suite.display_info()

    print("\nStandard Room Info:")
    standard.display_info()

    # Book the suite room
    print("\nBooking the suite room...")
    suite.book_room()

    # Check in the guest into the suite room
    print("Checking in the guest into the suite room...")
    suite.check_in()

    # Try to book the suite room again
    print("Trying to book the suite room again...")
    suite.book_room()

    # Check out the guest from the suite room
    print("Checking out the guest from the suite room...")
    suite.check_out()

    # Try to check in the guest into the suite room again
    print("Trying to check in the guest into the suite room again...")
    suite.check_in()

    # Display updated information
    print("\nUpdated Suite Room Info:")
    suite.display_info()

    print("\nUpdated Standard Room Info:")
    standard.display_info()

if __name__ == "__main__":
    main()
````

````python
class Member:
    _id_counter = 1  # Class-level attribute for unique ID generation

    def __init__(self, name, age, membership_type):
        self.__id = Member._id_counter
        Member._id_counter += 1
        self.name = name
        self.age = age
        self.membership_type = membership_type
        self.__membership_status = "Inactive"

    # Getter for ID
    def get_id(self):
        return self.__id

    # Method to register a new member
    def register_member(self):
        if self.__membership_status == "Inactive":
            self.__membership_status = "Active"
            print(f"Member {self.name} has been registered and activated.")
        else:
            print(f"Member {self.name} is already active.")

    # Method to renew membership
    def renew_membership(self):
        if self.__membership_status == "Active":
            print(f"Member {self.name}'s membership has been renewed.")
        else:
            print(f"Member {self.name} is not active. Please register the member first.")

    # Method to cancel membership
    def cancel_membership(self):
        if self.__membership_status == "Active":
            self.__membership_status = "Inactive"
            print(f"Member {self.name}'s membership has been cancelled.")
        else:
            print(f"Member {self.name} is already inactive.")

    # Method to display member information
    def display_info(self):
        status = self.__membership_status
        print(f"Member ID: {self.get_id()}")
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Membership Type: {self.membership_type}")
        print(f"Membership Status: {status}")

class FamilyMember(Member):
    def __init__(self, name, age, membership_type, family_size):
        super().__init__(name, age, membership_type)
        self.family_size = family_size

    # Method to display family member information
    def display_info(self):
        super().display_info()
        print(f"Family Size: {self.family_size}")

class IndividualMember(Member):
    def __init__(self, name, age, membership_type, personal_trainer):
        super().__init__(name, age, membership_type)
        self.personal_trainer = personal_trainer

    # Method to display individual member information
    def display_info(self):
        super().display_info()
        print(f"Personal Trainer: {'Yes' if self.personal_trainer else 'No'}")

def main():
    # Create instances of FamilyMember and IndividualMember
    family_member = FamilyMember("John Doe", 40, "Family", 4)
    individual_member = IndividualMember("Jane Smith", 28, "Individual", True)

    # Display initial information
    print("Family Member Info:")
    family_member.display_info()

    print("\nIndividual Member Info:")
    individual_member.display_info()

    # Register the family member
    print("\nRegistering the family member...")
    family_member.register_member()

    # Renew the membership of the individual member
    print("Renewing the individual member's membership...")
    individual_member.renew_membership()

    # Cancel the membership of the family member
    print("Cancelling the family member's membership...")
    family_member.cancel_membership()

    # Try to renew the membership of the family member after cancellation
    print("Trying to renew the family member's membership after cancellation...")
    family_member.renew_membership()

    # Display updated information
    print("\nUpdated Family Member Info:")
    family_member.display_info()

    print("\nUpdated Individual Member Info:")
    individual_member.display_info()

if __name__ == "__main__":
    main()
````

To create a Python program that reads a text file named "paragraph.txt" and counts the occurrences of each word in the paragraph, displaying the results in alphabetical order, you can follow these steps:

Open the file in read mode.

Read the content of the file.

Split the content into words and count the occurrences of each word.

Display the results in alphabetical order.

Here is the complete Python program:

````python
import string

def count_word_occurrences(file_path):
    """
    Count the occurrences of each word in the specified file and display the results in alphabetical order.

    :param file_path: The path to the file containing the paragraph.
    :return: A dictionary with words as keys and their counts as values.
    """
    word_counts = {}
    try:
        with open(file_path, 'r') as file:
            content = file.read().lower()  # Read the file content and convert to lowercase
            # Remove punctuation using str.translate and str.maketrans
            content = content.translate(str.maketrans('', '', string.punctuation))
            words = content.split()  # Split the content into words

            for word in words:
                if word in word_counts:
                    word_counts[word] += 1
                else:
                    word_counts[word] = 1

    except FileNotFoundError:
        print(f"The file {file_path} does not exist.")
    return word_counts

def display_word_counts(word_counts):
    """
    Display the word counts in alphabetical order.

    :param word_counts: A dictionary with words as keys and their counts as values.
    """
    for word in sorted(word_counts):
        print(f"{word}: {word_counts[word]}")

def main():
    # Path to the input file
    file_path = 'paragraph.txt'

    # Count the occurrences of each word in the file
    word_counts = count_word_occurrences(file_path)

    # Display the word counts in alphabetical order
    if word_counts:
        display_word_counts(word_counts)

if __name__ == "__main__":
    main()
````

### Measure of Central Tendency

**Definition**:
Measures of central tendency are statistical measures that describe the center point or typical value of a dataset. They provide a single value that represents the center of the data distribution.

**Types**:
1. **Mean (Arithmetic Average)**: The sum of all the data points divided by the number of data points.
2. **Median**: The middle value of a data set when the values are arranged in ascending or descending order. If the dataset has an even number of observations, the median is the average of the two middle numbers.
3. **Mode**: The most frequently occurring value in a dataset.

image.png

### Measure of Dispersion

**Definition**:
Measures of dispersion are statistical measures that describe the spread or variability of a dataset. They indicate how much the data points differ from the central tendency (mean, median, mode).

**Types**:
1. **Range**: The difference between the maximum and minimum values in a dataset.
2. **Variance**: The average of the squared differences from the mean.
3. **Standard Deviation**: The square root of the variance, providing a measure of spread in the same units as the data.
4. **Interquartile Range (IQR)**: The difference between the first quartile (Q1) and the third quartile (Q3), measuring the spread of the middle 50% of the data.

image.png

### Examples in Python

Here's how you can calculate these measures in Python using a dataset:

```python
import numpy as np
from scipy import stats

# Example dataset
data = [10, 15, 14, 18, 20, 30, 12, 25, 28, 10]

# Measure of Central Tendency
mean = np.mean(data)
median = np.median(data)
mode = stats.mode(data)[0][0]

# Measure of Dispersion
range_ = np.ptp(data)
variance = np.var(data, ddof=1)  # ddof=1 for sample variance
std_deviation = np.std(data, ddof=1)  # ddof=1 for sample standard deviation
q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)
iqr = q3 - q1

print(f"Mean: {mean}")
print(f"Median: {median}")
print(f"Mode: {mode}")
print(f"Range: {range_}")
print(f"Variance: {variance}")
print(f"Standard Deviation: {std_deviation}")
print(f"IQR: {iqr}")
```

### Skewness

**Definition**:
Skewness is a statistical measure that describes the asymmetry of the distribution of values in a dataset. It indicates whether the data values are skewed to the left (negative skew) or to the right (positive skew) or if they are symmetric (no skew).

**Types of Skewness**:
1. **Positive Skewness (Right-Skewed)**: The tail on the right side of the distribution is longer or fatter than the left side. Most of the data values are concentrated on the left with a few larger values stretching the tail to the right.
2. **Negative Skewness (Left-Skewed)**: The tail on the left side of the distribution is longer or fatter than the right side. Most of the data values are concentrated on the right with a few smaller values stretching the tail to the left.
3. **Zero Skewness (Symmetric Distribution)**: The distribution is symmetric. The left and right sides of the distribution are mirror images of each other.

### 1. Purpose

**Correlation**:
- The main goal of correlation is to measure the strength and direction of the linear relationship between two variables.
- It quantifies how strongly the two variables are related to each other. It indicates that the variables change together but does not imply causation.

**Regression**:
- The primary aim of regression is to model the relationship between a dependent variable and one or more independent variables. It seeks to predict or explain the dependent variable based on the independent variables.
- Regression provides an equation that describes the relationship between the variables, allowing for predictions and understanding of how changes in the independent variable(s) affect the dependent variable.

### 2. Directionality

**Correlation**:
- Correlation is symmetric, meaning the correlation between variable X and variable Y is the same as the correlation between variable Y and variable X.
- The variables are interchangeable; it does not matter which variable is labeled X and which is labeled Y.

**Regression**:
- Regression is asymmetric, meaning the dependent variable and the independent variable(s) have specific roles. The relationship is directional, from independent variable(s) to dependent variable.
- The variables are not interchangeable. Changing the dependent variable to an independent variable (or vice versa) would change the regression model entirely.

### 3. Measurement

**Correlation**:
- Correlation is measured by the correlation coefficient, typically Pearson's correlation coefficient (r), which ranges from -1 to 1.
  - r = 1: Perfect positive linear relationship.
  - r = -1: Perfect negative linear relationship.
  - r = 0: No linear relationship.

**Regression**:
- Regression results in a regression equation of the form Y = a + bX for simple linear regression, where:
  - Y is the dependent variable.
  - X is the independent variable.
  - a is the intercept.
  - b is the slope of the line (indicating the change in Y for a one-unit change in X).

### 4. Scope of Analysis

**Correlation**:
- Correlation only measures the degree of linear association between two variables. It does not imply any causation or the nature of the relationship beyond linearity.
- Correlation is a simpler analysis that provides a single summary statistic.

**Regression**:
- Regression can handle multiple variables and complex relationships. It provides a detailed analysis of how changes in the independent variable(s) affect the dependent variable.
- Regression is a more complex analysis that provides an equation, statistical significance, confidence intervals, and other diagnostic information.

Normal Distribution

Definition:

The normal distribution, also known as the Gaussian distribution, is a continuous probability distribution that is symmetric about its mean, showing that data near the mean are more frequent in occurrence than data far from the mean. In graph form, the normal distribution will appear as a bell curve.

Key Characteristics:

Bell-shaped Curve: The distribution has a single peak at the center and tapers off symmetrically on both sides.

Symmetry: The distribution is perfectly symmetric around the mean.
Mean, Median, and Mode: For a normal distribution, the mean, median, and mode are all equal and located at the center of the distribution.

Asymptotic: The tails of the distribution curve approach, but never touch, the horizontal axis.

68-95-99.7 Rule: About 68% of the data falls within one standard deviation of the mean, 95% falls within two standard deviations, and 99.7% falls within three standard deviations.

image.png



Four Assumptions of Normal Distribution

Assumption of Normality (Normality)

Explanation: The data should be approximately normally distributed. This means that when you plot the data, the resulting graph should have the bell-shaped curve of a normal distribution.

Validation: This can be visually checked using histograms, Q-Q plots, or more formally tested using statistical tests like the Shapiro-Wilk test, Anderson-Darling test, or the Kolmogorov-Smirnov test.

Assumption of Independence

Explanation: The observations should be independent of each other. This means that the occurrence of one observation does not influence the occurrence of another.

Validation: This assumption is more about the design of the study or experiment. Ensuring proper random sampling and avoiding systematic biases can help achieve independence.

Assumption of Homoscedasticity (Constant Variance)

Explanation: The variance of the errors should be constant across different values of the independent variable(s). This means the spread of the residuals should be roughly the same at all levels of the independent variable.

Validation: This can be checked by plotting the residuals versus fitted values. The plot should show a random scatter without any discernible pattern. Tests like the Breusch-Pagan test or White's test can also be used.

Assumption of Linearity

Explanation: There should be a linear relationship between the independent and dependent variables. This means the change in the dependent variable is proportional to the change in the independent variable.

Validation: This can be checked using scatter plots of the dependent variable versus the independent variable(s). If the relationship is linear, the scatter plot should show a straight-line pattern. Statistical tests and regression diagnostics can further confirm linearity.

Detailed Explanation of Assumptions

Normality

Importance: Many statistical tests, including t-tests, ANOVA, and linear regression, assume that the data or the residuals are normally distributed. Violations of normality can lead to incorrect conclusions.

How to Check:

Histogram: A bell-shaped curve indicates normality.

Q-Q Plot: If the data points lie approximately along the diagonal line, the data are normally distributed.

Statistical Tests: The Shapiro-Wilk test provides a formal assessment of normality. A p-value greater than 0.05 typically indicates that the data do not significantly deviate from normality.

Independence

Importance: Independence ensures that the observations provide unique information. Violations of independence can lead to underestimated standard errors and overestimated significance levels.

How to Check:

Design of Study: Ensuring random sampling or random assignment.

Time Series Data: Check for autocorrelation using the Durbin-Watson
statistic or autocorrelation plots.
Homoscedasticity

Importance: Constant variance ensures that the model's predictions are equally reliable across all levels of the independent variable. Heteroscedasticity can lead to inefficient estimates and biased standard errors.

How to Check:

Residual Plots: Plot the residuals against the fitted values. A random scatter indicates homoscedasticity, while a pattern (e.g., funnel shape) suggests heteroscedasticity.

Statistical Tests: The Breusch-Pagan test and White's test formally assess homoscedasticity.

Linearity

Importance: Linearity ensures that the estimated relationship accurately reflects the true relationship. Non-linearity can lead to biased estimates and incorrect conclusions.

How to Check:

Scatter Plots: Plot the dependent variable against each independent variable. A straight-line pattern indicates linearity.

Residual Plots: Plot the residuals against each independent variable. Random scatter suggests linearity, while patterns indicate non-linearity.
Visual Representation in Python

Let's create visualizations to check these assumptions using Python:

````python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from scipy import stats

# Generate sample data
np.random.seed(0)
data = np.random.normal(loc=0, scale=1, size=1000)

# Histogram to check Normality
plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
sns.histplot(data, kde=True)
plt.title('Histogram')

# Q-Q Plot to check Normality
plt.subplot(2, 2, 2)
sm.qqplot(data, line='45')
plt.title('Q-Q Plot')

# Generate data for other assumptions
x = np.random.normal(loc=5, scale=2, size=100)
y = 2 * x + np.random.normal(loc=0, scale=2, size=100)

# Scatter plot to check Linearity
plt.subplot(2, 2, 3)
plt.scatter(x, y)
plt.title('Scatter Plot for Linearity')
plt.xlabel('Independent Variable')
plt.ylabel('Dependent Variable')

# Residual plot to check Homoscedasticity
model = sm.OLS(y, sm.add_constant(x)).fit()
residuals = model.resid
plt.subplot(2, 2, 4)
plt.scatter(model.fittedvalues, residuals)
plt.hlines(0, min(model.fittedvalues), max(model.fittedvalues), colors='red')
plt.title('Residual Plot for Homoscedasticity')
plt.xlabel('Fitted Values')
plt.ylabel('Residuals')

plt.tight_layout()
plt.show()

# Statistical tests
shapiro_test = stats.shapiro(data)
breusch_pagan_test = sm.stats.diagnostic.het_breuschpagan(residuals, model.model.exog)
durbin_watson_test = sm.stats.durbin_watson(residuals)

print(f"Shapiro-Wilk test for normality: {shapiro_test}")
print(f"Breusch-Pagan test for homoscedasticity: {breusch_pagan_test}")
print(f"Durbin-Watson test for independence: {durbin_watson_test}")
````

Skewness and Kurtosis:

#### 5. Create a Systemd Service File for Gunicorn

```

WorkingDirectory=/home/your_username/myflaskapp
Environment="PATH=/home/your_username/myflaskapp/venv/bin"
ExecStart=/home/your_username/myflaskapp/venv/bin/gunicorn --workers 3 --bind unix:myflaskapp.sock -m 007 app:app

[Install]
WantedBy=multi-user.target
```

Replace `your_username` with your actual username and `/home/your_username/myflaskapp` with the path to your Flask app directory.

Replace `your_server_domain_or_IP` with your server's domain name or IP address, and `/home/your_username/myflaskapp/myflaskapp.sock` with the path to your Gunicorn socket file.

Creating a fully functional web application using Flask and MongoDB with signup and signin pages requires several steps. Below is a step-by-step guide on how to achieve this. This example uses `flask-pymongo` to interact with MongoDB and `flask-wtf` for form handling and validation.

from flask_wtf import FlaskForm
from wtforms import StringField, PasswordField, SubmitField

app.config['MONGO_URI'] = 'mongodb://localhost:27017/mydatabase'

# User registration form
class RegistrationForm(FlaskForm):
    email = StringField('Email', validators=[DataRequired(), Email()])
    password = PasswordField('Password', validators=[DataRequired(), Length(min=6)])
    confirm_password = PasswordField('Confirm Password', validators=[DataRequired(), EqualTo('password')])
    submit = SubmitField('Sign Up')

# User login form
class LoginForm(FlaskForm):
    email = StringField('Email', validators=[DataRequired(), Email()])
    password = PasswordField('Password', validators=[DataRequired()])

@app.route('/signup', methods=['GET', 'POST'])
def signup():
    form = RegistrationForm()
    if form.validate_on_submit():
        email = form.email.data
        hashed_password = bcrypt.generate_password_hash(form.password.data).decode('utf-8')
        mongo.db.users.insert_one({'email': email, 'password': hashed_password})
        flash('Account created successfully! Please log in.', 'success')
        return redirect(url_for('signin'))
    return render_template('signup.html', form=form)

@app.route('/signin', methods=['GET', 'POST'])
def signin():
    form = LoginForm()
    if form.validate_on_submit():
        user = mongo.db.users.find_one({'email': form.email.data})
        if user and bcrypt.check_password_hash(user['password'], form.password.data):
            session['user_id'] = user['_id']
            session['email'] = user['email']
            flash('Login successful!', 'success')
            return redirect(url_for('hello'))
        else:
            flash('Login unsuccessful. Please check your email and password.', 'danger')
    return render_template('signin.html', form=form)

@app.route('/hello')
def hello():
    if 'user_id' not in session:
        return redirect(url_for('signin'))
    return render_template('hello.html', email=session['email'])

@app.route('/logout')
def logout():
    session.pop('user_id', None)
    session.pop('email', None)
    flash('You have been logged out.', 'success')
    return redirect(url_for('signin'))
