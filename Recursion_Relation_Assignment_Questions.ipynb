T(n)=T(n−1)+c

```python
def T(n, c):
    if n == 0:
        return 0  # Assume T(0) = 0 for simplicity
    else:
        return T(n - 1, c) + c

# Example usage
n = 4
c = 2
result = T(n, c)
print(f"The value of T({n}) is: {result}")
```

B. T(n)=2T(n/2)+n

```python
def T(n):
    if n == 1:
        return 1  # Base case
    else:
        return 2 * T(n // 2) + n

# Example usage
n = 8  # Example value
result = T(n)
print(f"The value of T({n}) is: {result}")
```

C. T(n)=2T(n/2)+c

```python
def T(n, c):
    if n == 1:
        return c  # Base case
    else:
        return 2 * T(n // 2, c) + c

# Example usage
n = 8  # Example value
c = 3  # Example constant
result = T(n, c)
print(f"The value of T({n}) is: {result}")
```
